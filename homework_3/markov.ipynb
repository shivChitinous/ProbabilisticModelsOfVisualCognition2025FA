# Markov Transition Probabilities

```python
import numpy as np
import scipy as sp
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
def transition_matrix_pandas(x):
    s = pd.Series(list(x))
    pairs = pd.crosstab(s.shift(1).dropna(), s[1:])

    states = sorted(set(s))
    pairs = pairs.reindex(index=states, columns=states, fill_value=0)

    T = pairs.div(pairs.sum(axis=1), axis=0).rename_axis('t-1').rename_axis('t', axis=1)
    return T, states
```

```python
x = [0, 1, 1]
T, states = transition_matrix_pandas(x)
T
```
